Stemming with PorterStemmer (less accurate when finding the stem of the word)

Porter's Stemmer is one of the most popular and widely used stemming algorithms. Proposed in 1980 by Martin Porter, this stemmer works by applying a series of rules to remove common suffixes from English words. It is well-known for its simplicity, speed and reliability. However, the stemmed output is not guaranteed to be a meaningful word and its applications are limited to the English language.

```python
from nltk.stem import PorterStemmer

words = ["running", "jumps", "happily", "running", "foxes"]

porter_stemmer = PorterStemmer()

stemmed_words = [porter_stemmer.stem(word) for word in words]

print("Original words:", words)
print("Stemmed words:", stemmed_words)
```

Snowball Stemmer (much accurate than the porterstemmer)

The Snowball Stemmer is an enhanced version of the Porter Stemmer which was introduced by Martin Porter as well. It is referred to as Porter2 and is faster and more aggressive than its predecessor. One of the key advantages of this is that it supports multiple languages, making it a multilingual stemmer.

```python
from nltk.stem import SnowballStemmer

snowball_stemmer = SnowballStemmer(language='english')
words = ["running", "jumps", "happily", "running", "foxes"]

stemmed_words = [snowball_stemmer.stem(word) for word in words]

print("Original words:", words)
print("Stemmed words:", stemmed_words)
```

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng') ## required for english sentences
```

```python
## apply the text pre-processing steps like token->text removal(stopwords) -> stemming -> lemmitization
import nltk
from nltk.stem import SnowballStemmer
from nltk.stem import WordNetLemmatizer

paragraph = """The quick brown fox jumps over the lazy dog.""" 

# 1.tokenization of the corpos into sentences
sentences= nltk.sent_tokenize(paragraph)
stemmer = SnowballStemmer(language='english')
lemmitization = WordNetLemmatizer()

for i in range(len(sentences)):
    ## break down each sentence into word(word tokenization)
    words = nltk.word_tokenize(sentences[i]) ## retrieving the single sentence from the list of sentences and breaking down(tokenization into words)
    # 2. apply the snowball stemming and then apply the lemmitization to regulirized the words
    # 4. filter the common words(verbs) using the stopwords 
    ## below pos stands for part of speech tag used in the snowball stemmer
    stemmed_words = [lemmitization.lemmatize(stemmer.stem(word),pos='v') for word in words if word not in set(stopwords.words('english'))] ## filter words from the stopword of english
    sentences[i]=' '.join(stemmed_words) ## conerting all the words into sentences
```

```python
## we will find the POST tag (part of speech)
para = """The quick brown fox jumps over the lazy dog."""
para_list=para.split()
for i in range(len(para_list)):
    words = nltk.word_tokenize(para_list[i])
    words=[word for word in words if word not in set(stopwords.words('english'))]
    pos_tag=nltk.pos_tag(words)
    print(pos_tag)

print("==================")

print(nltk.pos_tag("The quick brown fox jumps over the lazy dog.".split()))
    

```
